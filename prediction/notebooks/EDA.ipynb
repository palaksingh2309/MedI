# Exploratory Data Analysis (EDA) - MedIntel Disease Symptom Prediction

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
csv_path = '../datasets/symptoms.csv'
df = pd.read_csv(csv_path)
print(f'Dataset dimensions: {df.shape[0]} rows, {df.shape[1]} columns')
df.head()
```

```python
# Number of diseases and unique symptom list
diseases = df['disease'].unique()
symptoms = df.columns.drop('disease')
print(f'Number of unique diseases: {len(diseases)}')
print(f'Number of unique symptoms: {len(symptoms)}')
```

```python
# Plot the distribution of diseases to check for class imbalance
plt.figure(figsize=(12, 6))
sns.countplot(y='disease', data=df, order=df['disease'].value_counts().index, palette='viridis')
plt.title('Disease Distribution in Dataset')
plt.xlabel('Count')
plt.ylabel('Disease')
plt.tight_layout()
plt.show()
```

```python
# Analyze average number of symptoms per disease
df['symptom_count'] = df[symptoms].sum(axis=1)
avg_symptoms = df.groupby('disease')['symptom_count'].mean().reset_index()

plt.figure(figsize=(12, 6))
sns.barplot(y='disease', x='symptom_count', data=avg_symptoms.sort_values(by='symptom_count', ascending=False), palette='magma')
plt.title('Average Number of Active Symptoms per Disease')
plt.xlabel('Average Symptom Count')
plt.ylabel('Disease')
plt.tight_layout()
plt.show()
```

```python
# Identify the top 15 most common symptoms across all records
symptom_frequencies = df[symptoms].sum().sort_values(ascending=False).head(15).reset_index()
symptom_frequencies.columns = ['symptom', 'frequency']

plt.figure(figsize=(10, 5))
sns.barplot(x='frequency', y='symptom', data=symptom_frequencies, palette='coolwarm')
plt.title('Top 15 Most Common Symptoms')
plt.xlabel('Occurrence Count')
plt.ylabel('Symptom')
plt.tight_layout()
plt.show()
```
